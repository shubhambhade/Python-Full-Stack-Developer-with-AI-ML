```python
# Addition program - POP
number1 = 10
number2 = 20 
sum = number1 + number2
print("Addition of two numbers : ", sum)
```

```python
# Addition program - OOP
class Addition():
    def add(self):
        number1 = 10
        number2 = 20
        sum = number1 + number2
        print("Addition of two numbers : ", sum)

addition = Addition()
addition.add()
```

```python
# functional programming

def addition():
    number1 = 10
    number2 = 20
    sum =  number1 + number2
    print("Addition of two numbers : ", sum)

addition()
```
