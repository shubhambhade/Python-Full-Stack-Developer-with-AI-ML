```python
# check pallindrome

s="madam"

start = 0
end = len(s)-1

is_pallindrome = True

while start < end:
    if s[start] != s[end]:
        print("string not pallindrome")
        is_pallindrome = False

    else:
        start += 1
        end += -1

if is_pallindrome:
    print("string is pallindrome")

    
```

```python
# Count Vowels and Consonants
s = 'python is programming language'

vowels = []
consonanats = []

for ch in s:
    if ch in 'oeiou':
        vowels.append(ch)
    else:
        consonanats.append(ch)

print("vowels count : ", len(vowels))
print("consonanats count : ", len(consonanats))
```

```python
# Count Character Frequency
s='python programming'
freq={}

for ch in s:
    if ch not in freq:
        freq[ch] = 1
    else:
        freq[ch] += 1

print(freq)
```

```python
# Find Duplicate Characters
s='python pro'
freq = {}

for ch in s:
    if ch not in freq:
        freq[ch] = 1
    else:
        freq[ch]  += 1

for key , val in freq.items():
    if val > 1:
        print(key)
```
