```python
# wap to find out numerical sum of all unique characters presents inside string.
s = "aassqbjd"
unique_character = ""
for ch in s:
    if ch not in unique_character:
        unique_character += ch
print(unique_character)

sum_of_characters = 0
for ch in unique_character:
    sum_of_characters += ord(ch)

print("Sum of charcters : ", sum_of_characters)
```

```python
# wap to converting capital letter to small letters
# A = 65
# a = 97

s = "ABCDEFG"
small_letter = "" 
for ch in s:
    convert_capital_to_small_letter = ord(ch) + 32
    small_letter += chr(convert_capital_to_small_letter)
    convert_capital_to_small_letter = 0
print("after converting capital to small letter string is : ",small_letter)
```

```python
# wap to print index position of every repitition of character in string.
s = "most of the people are lazy to come to class"
d = {}

for i in range(len(s)):
    ch  = s[i]
    if ch not in d:
        d[ch] = [i]
    else:
        d[ch].append(i)

for key, value in d.items():
    print(f'{key} : {value}')
```

```python
# write a program to create 2 different list alternate character present in string.
s = "most of the people are lazy to come to class"
odd_position_character_list = []
even_position_character_list = []
for i in range(len(s)):
    if i % 2 == 0:
        even_position_character_list.append(s[i])
    else:
        odd_position_character_list.append(s[i])
print("Even position characters : ",even_position_character_list)
print("odd position characters : ",odd_position_character_list)
```

```python
# wap to reverse the words in the string. 
s = "hello every one good morning"
words = s.split()

l = len(words) - 1
revesre_words_in_string = ""
while l >= 0:
    revesre_words_in_string += words[l] +" "
    l -= 1

print(revesre_words_in_string.rstrip())
```

```python
s = "hello every one good morning"
words = s.split()
words = words[::-1]
revesre_words_in_string = " ".join(words)
print(revesre_words_in_string)
```

```python
# wap to reverse the words in string in place 
s = "hello every one good morning"
words = s.split()
ans = "" 
for word in words:
    ans += word[::-1] + " "
print(ans)
```

```python
#write a program
s1 = "sham"
s2 = "ram"
# output :alternate character both string : srhaamm

ans="" 
i = 0
while i < len(s1) or i < len(s2):
    if i < len(s1):
        ans += s1[i]

    if i < len(s2):
        ans += s2[i]

    i += 1

print(ans)
```
